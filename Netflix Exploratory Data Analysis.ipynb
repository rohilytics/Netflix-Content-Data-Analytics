#### 📌Q. How are contents distributed in Netflix Platform ?

```python
ntype_gp = nxd.groupby('type')['show_id'].nunique().reset_index()
ntype_gp
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Netflix Contents Distribution (Pie)", "Netflix Contents Distribution (Bar)"),
    specs=[[{'type':'domain'}, {'type':'xy'}]]
)

# Pie chart
fig.add_trace(
    go.Pie(
        labels=ntype_gp['type'],      # <-- use the 'type' column for labels
        values=ntype_gp['show_id'],
        hole=0,
        marker=dict(colors=['#cb4335', '#dedede']),
        pull=[0.08, 0],
        textinfo='percent+label',
        textfont=dict(color='black')
    ),
    row=1, col=1
)

# Bar chart
fig.add_trace(
    go.Bar(
        x= ntype_gp['type'],
        y= ntype_gp['show_id'],
        marker_color=['#cb4335', '#dedede'],
        text= ntype_gp['show_id'],
        textposition='outside'
    ),
    row=1, col=2
)

fig.update_layout(
    title_text="Netflix Contents Distribution",
    showlegend=False,
    height=600,
    width=1200
)
fig.update_yaxes(showticklabels=False, title_text='', row=1, col=2)
fig.show()
```

#### 👉Insights : 

- we can clearly interpret that nearly *70%* contents are `Movies` whereas *30%* are `Tvshows` contents in **Netflix** content library.


#### 📌Q. Outliers check: 

```python
from plotly.subplots import make_subplots

import plotly.graph_objects as go

# Prepare data for violin and box plots
types = nxd['type'].unique()
colors = ['#cb4335', "#A2A1A1"]

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Violin Distribution", "Box & Whisker Distribution"),
    shared_yaxes=True
)

# Violin plots
for idx, t in enumerate(types):
    fig.add_trace(
        go.Violin(
            x=nxd[nxd['type'] == t]['type'],
            y=nxd[nxd['type'] == t]['release_year'],
            name=t,
            line_color=colors[idx],
            fillcolor=colors[idx],
            opacity=0.7,
            box_visible=True,
            meanline_visible=True,
            showlegend=False
        ),
        row=1, col=1
    )

# Box plots
for idx, t in enumerate(types):
    fig.add_trace(
        go.Box(
            x=nxd[nxd['type'] == t]['type'],
            y=nxd[nxd['type'] == t]['release_year'],
            name=t,
            marker_color=colors[idx],
            boxmean=True,
            showlegend=False
        ),
        row=1, col=2
    )

fig.update_layout(
    height=800,
    width=1500,
    title_text="Netflix Contents Distribution",
    margin=dict(t=80),
)

fig.update_xaxes(title_text="", row=1, col=1)
fig.update_xaxes(title_text="", row=1, col=2)
fig.update_yaxes(title_text="Release Year", row=1, col=1)
fig.update_yaxes(title_text="", row=1, col=2)

fig.show()
```

#### 👉Insights : 

- Here we can easily identify that the contents released before 1960 are the outliers.

#### 📌Q. In which year maximum contents got released ?

```python
import plotly.express as px

# Filter for years 1998 to 2021
filtered = nxd[(nxd['release_year'] >= 1998) & (nxd['release_year'] <= 2021)]

# Group by year and type, count unique show_id
year_type_counts = (
    filtered.groupby(['release_year', 'type'])['show_id']
    .nunique()
    .reset_index(name='content_count')
    .sort_values('release_year')
)

# Plot
fig = px.line(
    year_type_counts,
    x='release_year',
    y='content_count',
    color='type',
    markers=True,
    title='Content Released Each Year (2000–2021)',
    labels={'release_year': 'Year Added', 'content_count': 'Number of Contents', 'type': 'Content Type'},
    color_discrete_map={'Movie': '#CB4335', 'TV Show': '#A2A1A1'},
    line_shape = 'spline',
)

fig.update_layout(
    width=1200,
    height=700,
    xaxis=dict(
        tickmode='linear',
        dtick=1
    )
)
fig.show()
```

```python
import plotly.express as px

# Filter for years 1998 to 2021
filtered = nxd[(nxd['year_added'] >= 1998) & (nxd['year_added'] <= 2021)]

# Group by year and type, count unique show_id
year_type_counts = (
    filtered.groupby(['year_added', 'type'])['show_id']
    .nunique()
    .reset_index(name='content_count')
    .sort_values('year_added')
)

# Plot
fig = px.line(
    year_type_counts,
    x='year_added',
    y='content_count',
    color='type',
    markers=True,
    title='Content Added on Netflix Each Year (2000–2021)',
    labels={'year_added': 'Year Added', 'content_count': 'Number of Contents', 'type': 'Content Type'},
    color_discrete_map={'Movie': '#CB4335', 'TV Show': '#A2A1A1'},
    line_shape = 'spline',
)

fig.update_layout(
    width=1200,
    height=700,
    xaxis=dict(
        tickmode='linear',
        dtick=1
    )
)
fig.show()
```

#### 👉Insights :

- Between 2015 and 2016, Netflix saw a drastic increase in the number of movies and TV shows releases, marking a significant jump in content production or acquisitions. This surge was followed by a steady upward trend until 2019, reflecting continued investment in expanding the platform's movie catalog.

- However, starting from 2019, the number of movies released dropped sharply—likely due to the global impact of the COVID-19 pandemic on film production and distribution. This disruption appears as a clear dip in the trend, contrasting with the prior growth pattern. 

#### 📌What are the top 15 countries consumption of Movies and TV shows ? 

```python
import plotly.express as px

# Calculate top 15 countries by unique movie count, excluding 'Unknown'
cwm = (
    mvd[mvd['country'] != 'Unknown']
    .groupby('country')['show_id']
    .nunique()
    .nlargest(10)
    .sort_values(ascending=True)  # ascending for reversed y-axis
    .reset_index()
    .rename(columns={'show_id': 'movie_count'})
)

# Plot horizontal bar chart
fig = px.bar(
    cwm,
    x='movie_count',
    y='country',
    orientation='h',
    title='Top 15 Countries by Unique Netflix Movies',
    labels={'movie_count': 'Number of Movies', 'country': 'Country'},
    color_discrete_sequence=['#CB4335'],
    category_orders={'country': cwm['country'].tolist()}
)

fig.update_layout(
    width=1200,
    height=700,
    yaxis=dict(autorange='reversed')
)

fig.show()

```

```python
# Calculate top 15 countries by unique movie count, excluding 'Unknown'
ctv = (
    tvd[tvd['country'] != 'Unknown']
    .groupby('country')['show_id']
    .nunique()
    .nlargest(10)
    .sort_values(ascending=True)  # ascending for reversed y-axis
    .reset_index()
    .rename(columns={'show_id': 'tvshows_count'})
)

# Plot horizontal bar chart
fig = px.bar(
    ctv,
    x='tvshows_count',
    y='country',
    orientation='h',
    title='Top 15 Countries by Unique Netflix TV Shows',
    labels={'tvshows_count': 'Number of Movies', 'country': 'Country'},
    color_discrete_sequence=['#CB4335'],
    category_orders={'country': ctv['country'].tolist()}
)

fig.update_layout(
    width=1200,
    height=700,
    yaxis=dict(autorange='reversed')
)

fig.show()
```

```python
common_countries = set(cwm['country']).intersection(ctv['country'])

# Filter both to common countries
cwm_aligned = cwm[cwm['country'].isin(common_countries)].reset_index(drop=True)

ctv_aligned = ctv[ctv['country'].isin(common_countries)]

# Reorder ctv_aligned to the same country order as cwm_aligned
ctv_aligned = ctv_aligned.set_index('country').loc[cwm_aligned['country']].reset_index()

# Now check order again:
print(list(cwm_aligned['country']) == list(ctv_aligned['country']))  # should be True
```

```python
import plotly.graph_objects as go

top_n = len(cwm_aligned)
indices = list(range(top_n))

fig = go.Figure()

# Movies bars
fig.add_trace(go.Bar(
    x=indices,
    y=cwm_aligned['movie_count'],
    name='Movies',
    marker_color='#cb4335',
    width=0.4,
    text=cwm_aligned['movie_count'],
    textposition='outside'
))

# TV Shows bars
fig.add_trace(go.Bar(
    x=indices,
    y=ctv_aligned['tvshows_count'],
    name='TV Shows',
    marker_color='#A2A1A1',
    width=0.4,
    text=ctv_aligned['tvshows_count'],
    textposition='outside'
))

fig.update_layout(
    title='Countries Consuming Movies & TV Shows',
    xaxis=dict(
        tickmode='array',
        tickvals=indices,
        ticktext=cwm_aligned['country'],
        title='Country'
    ),
    yaxis_title='Content Count',
    barmode='group',
    legend=dict(
        orientation='h',   # horizontal legend
        yanchor='bottom',
        y=1.08,            # move legend above plot
        xanchor='center',
        x=0.5
    ),
    width=1200,
    height=600,
    font=dict(family='sans-serif', size=13),
    margin=dict(t=80, l=60, r=30, b=80)
)

fig.show()
```

#### 👉Insights :

- The top 5 countries with the highest count of Movies and TV Shows are aiding in recognizing the key players in TV show production and Movies production.
- We can infer that `US` , `India` , `UK` are the top entertainment consumers while other countries significantly contribute to the OTT content library.
